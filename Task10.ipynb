# Task 10: Wasserstein GAN with Gradient Penalty (WGAN-GP)
Implementing Lipschitz-1 gradient constraint penalty for adversarial generator-critic training.


```python
import torch
from wgan_gp_synthesis import Generator, Critic, compute_gradient_penalty
print('WGAN-GP initialized.')
```

```python
# Run 1 step of Critic and Generator loss calculations
z = torch.randn(16, 64)
real = torch.randn(16, 784)
G = Generator()
C = Critic()
fake = G(z)

critic_real = C(real).mean()
critic_fake = C(fake).mean()
gp = compute_gradient_penalty(C, real, fake)

critic_loss = critic_fake - critic_real + 10.0 * gp
print(f'Critic Real: {critic_real.item():.4f} | Critic Fake: {critic_fake.item():.4f}')
print(f'Gradient Penalty: {gp.item():.4f} | Total Critic Loss: {critic_loss.item():.4f}')
```
