# Task 3: Vectorized Convolutional Forward Pass via im2col
Transforming sliding window convolutions into a single efficient matrix multiplication (GEMM).


```python
import numpy as np
from im2col_conv import Conv2D_im2col, im2col
np.random.seed(42)
print('NumPy im2col conv environment ready.')
```

```python
# Input tensor: Batch=8, Channels=3, Height=32, Width=32
X = np.random.randn(8, 3, 32, 32)
conv = Conv2D_im2col(in_channels=3, out_channels=16, kernel_size=3, padding=1, stride=1)
out = conv.forward(X)
print(f'Input shape: {X.shape} -> Output shape: {out.shape}')
```

```python
# Verify output statistics
print(f'Output Mean: {out.mean():.4f} | Output Std: {out.std():.4f}')
```
