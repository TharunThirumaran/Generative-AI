# Task 5: Custom Batch Normalization and Layer Normalization
Building structural normalization layers from scratch and verifying normalization properties.


```python
import torch
from custom_normalization_layers import CustomBatchNorm1d, CustomLayerNorm
print('PyTorch Normalization module loaded.')
```

```python
# Test Batch Normalization across batch dimension (32 samples, 10 features)
x = torch.randn(32, 10)
bn = CustomBatchNorm1d(10)
bn_out = bn.forward(x)
print(f'BatchNorm Output Mean: {bn_out.mean(dim=0).abs().max():.6f} (approx 0)')
print(f'BatchNorm Output Std:  {bn_out.std(dim=0).mean():.4f} (approx 1)')
```

```python
# Test Layer Normalization across feature dimension per sample
ln = CustomLayerNorm(10)
ln_out = ln.forward(x)
print(f'LayerNorm Sample 0 Mean: {ln_out[0].mean():.6f} | Std: {ln_out[0].std():.4f}')
```
